```python
import os
import cv2
import numpy as np

print("Generating Synthetic Metal Washer Dataset...")

# Create directory structure
folders = ["train/good", "test/good", "test/scratch", "test/dent"]
for f in folders:
    os.makedirs(f"washers/{f}", exist_ok=True)

def create_washer():
    # Base background (dark gray conveyor belt)
    img = np.ones((128, 128, 3), dtype=np.uint8) * 40

    # Outer circle (metal washer body)
    cv2.circle(img, (64, 64), 45, (170, 175, 180), -1)

    # Inner circle (the hole)
    cv2.circle(img, (64, 64), 18, (40, 40, 40), -1)

    # Add metallic grain/noise to make it look real
    noise = np.random.randint(-15, 15, (128, 128, 3))
    img = np.clip(img.astype(np.int16) + noise, 0, 255).astype(np.uint8)

    # Add a slight blur to smooth the edges
    img = cv2.GaussianBlur(img, (3, 3), 0)
    return img

# 1. Generate 200 Perfect Washers for Training
for i in range(200):
    cv2.imwrite(f"washers/train/good/{i}.png", create_washer())

# 2. Generate 20 Perfect Washers for Testing
for i in range(20):
    cv2.imwrite(f"washers/test/good/{i}.png", create_washer())

# 3. Generate 20 Defective Washers (Scratches)
for i in range(20):
    img = create_washer()
    # Draw a random white scratch mark
    x1, y1 = np.random.randint(30, 98, 2)
    x2, y2 = x1 + np.random.randint(-25, 25), y1 + np.random.randint(-25, 25)
    cv2.line(img, (x1, y1), (x2, y2), (230, 230, 255), np.random.randint(1, 3))
    cv2.imwrite(f"washers/test/scratch/{i}.png", img)

# 4. Generate 20 Defective Washers (Dents/Contamination)
for i in range(20):
    img = create_washer()
    # Draw a random dark dent/blob
    x, y = np.random.randint(35, 93, 2)
    cv2.circle(img, (x, y), np.random.randint(3, 7), (70, 70, 70), -1)
    cv2.imwrite(f"washers/test/dent/{i}.png", img)

print("Dataset generated successfully in the 'washers' folder!")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import transforms
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import os

class WasherDataset(Dataset):
    def __init__(self, dir_path):
        self.dir_path = dir_path
        self.img_names = [f for f in os.listdir(dir_path) if f.endswith('.png')]
        self.transform = transforms.Compose([
            transforms.Resize((128, 128)),
            transforms.ToTensor()
        ])

    def __len__(self):
        return len(self.img_names)

    def __getitem__(self, idx):
        img_path = os.path.join(self.dir_path, self.img_names[idx])
        image = Image.open(img_path).convert("RGB")
        return self.transform(image)

class Autoencoder(nn.Module):
    def __init__(self):
        super(Autoencoder, self).__init__()
        self.encoder = nn.Sequential(
            nn.Conv2d(3, 16, 3, stride=2, padding=1),
            nn.ReLU(),
            nn.Conv2d(16, 32, 3, stride=2, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 64, 3, stride=2, padding=1),
            nn.ReLU()
        )
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(64, 32, 3, stride=2, padding=1, output_padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(32, 16, 3, stride=2, padding=1, output_padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(16, 3, 3, stride=2, padding=1, output_padding=1),
            nn.Sigmoid()
        )

    def forward(self, x):
        x = self.encoder(x)
        x = self.decoder(x)
        return x

dataset = WasherDataset("washers/train/good")
dataloader = DataLoader(dataset, batch_size=16, shuffle=True)

model = Autoencoder()
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

for epoch in range(20):
    for data in dataloader:
        optimizer.zero_grad()
        outputs = model(data)
        loss = criterion(outputs, data)
        loss.backward()
        optimizer.step()

torch.save(model.state_dict(), "autoencoder.pth")
```

```python
import gradio as gr
import torch
import torch.nn as nn
from torchvision import transforms
from PIL import Image
import numpy as np
import cv2

class Autoencoder(nn.Module):
    def __init__(self):
        super(Autoencoder, self).__init__()
        self.encoder = nn.Sequential(
            nn.Conv2d(3, 16, 3, stride=2, padding=1),
            nn.ReLU(),
            nn.Conv2d(16, 32, 3, stride=2, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 64, 3, stride=2, padding=1),
            nn.ReLU()
        )
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(64, 32, 3, stride=2, padding=1, output_padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(32, 16, 3, stride=2, padding=1, output_padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(16, 3, 3, stride=2, padding=1, output_padding=1),
            nn.Sigmoid()
        )

    def forward(self, x):
        x = self.encoder(x)
        x = self.decoder(x)
        return x

model = Autoencoder()
model.load_state_dict(torch.load("autoencoder.pth", map_location=torch.device('cpu')))
model.eval()

transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor()
])

def detect_defect(image):
    img_tensor = transform(image).unsqueeze(0)

    with torch.no_grad():
        reconstruction = model(img_tensor)

    diff = torch.abs(img_tensor - reconstruction).squeeze().numpy()
    diff = np.transpose(diff, (1, 2, 0))
    gray_diff = cv2.cvtColor(diff, cv2.COLOR_RGB2GRAY)

    gray_diff = (gray_diff * 255).astype(np.uint8)
    heatmap = cv2.applyColorMap(gray_diff, cv2.COLORMAP_JET)
    heatmap = cv2.cvtColor(heatmap, cv2.COLOR_BGR2RGB)

    anomaly_score = float(np.mean(gray_diff))

    if anomaly_score > 3.0:
        status = "DEFECT DETECTED - REJECT"
    else:
        status = "PASSED - NORMAL"

    return heatmap, f"{anomaly_score:.2f}", status

interface = gr.Interface(
    fn=detect_defect,
    inputs=gr.Image(type="pil", label="Upload Factory Component"),
    outputs=[
        gr.Image(label="Defect Heatmap"),
        gr.Textbox(label="Anomaly Score"),
        gr.Textbox(label="System Status")
    ],
    title="Edge AI Industrial Defect Monitor",
    description="Upload a component image. The model runs locally to detect anomalies without cloud processing."
)

interface.launch(share=True)
```
